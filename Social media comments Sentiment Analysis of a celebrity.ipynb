```python
!pip install google-api-python-client pandas

from googleapiclient.discovery import build
import pandas as pd
import time

API_KEY = "AIzaSyA16iYayKT-hPvOFflkJlO6CpGy8OT30EE"

video_ids = [
    "https://www.youtube.com/watch?v=ng7QdRXtQrg",
    "https://www.youtube.com/watch?v=Fo_lDTDSO-M",
    "https://www.youtube.com/watch?v=w5e1rK4xZps",
    "https://www.youtube.com/watch?v=SZI-KuzL4dg",
    "https://www.youtube.com/watch?v=sGjxPXekFEs&list=RDsGjxPXekFEs&start_radio=1",
    "https://www.youtube.com/watch?v=7Dg3UZU-l_k",
    "https://www.youtube.com/watch?v=-JmNKGfFj7w"
]

youtube = build("youtube", "v3", developerKey=API_KEY)

rows = []

def fetch_comments_for_video(video_id, max_pages=5):
    next_page_token = None
    page_count = 0

    while page_count < max_pages:
        request = youtube.commentThreads().list(
            part="snippet,replies",
            videoId=video_id,
            maxResults=100,
            pageToken=next_page_token,
            textFormat="plainText"
        )

        response = request.execute()

        for item in response["items"]:
            top = item["snippet"]["topLevelComment"]["snippet"]
            top_comment_id = item["snippet"]["topLevelComment"]["id"]

            rows.append({
                "video_id": video_id,
                "comment_id": top_comment_id,
                "parent_id": None,
                "level": 1,
                "author": top.get("authorDisplayName"),
                "text": top.get("textDisplay"),
                "likes": top.get("likeCount"),
                "published_at": top.get("publishedAt"),
                "reply_count": item["snippet"].get("totalReplyCount", 0)
            })

            if "replies" in item:
                for reply in item["replies"]["comments"]:
                    r = reply["snippet"]
                    rows.append({
                        "video_id": video_id,
                        "comment_id": reply["id"],
                        "parent_id": r.get("parentId"),
                        "level": 2,
                        "author": r.get("authorDisplayName"),
                        "text": r.get("textDisplay"),
                        "likes": r.get("likeCount"),
                        "published_at": r.get("publishedAt"),
                        "reply_count": 0
                    })

        next_page_token = response.get("nextPageToken")
        page_count += 1

        if not next_page_token:
            break

        time.sleep(1)

for vid in video_ids:
    try:
        # Correctly extract the video ID from the URL
        parsed_video_id = vid.split("v=")[1].split("&")[0] if "v=" in vid else vid
        fetch_comments_for_video(parsed_video_id, max_pages=5)
        print(f"Done: {vid}")
    except Exception as e:
        print(f"Skipped {vid}: {e}")

df = pd.DataFrame(rows)
df.to_csv("jim_carrey_yt_com.csv", index=False)
print(df.shape)
df.head()
```
